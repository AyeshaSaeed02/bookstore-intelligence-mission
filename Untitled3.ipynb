```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin

base_url = "https://books.toscrape.com/"
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

books = []

for page_number in range(1, 4):
    if page_number == 1:
        url = base_url
    else:
        url = f"{base_url}catalogue/page-{page_number}.html"

    response = requests.get(url, timeout=20)
    response.raise_for_status()
    response.encoding = "utf-8"

    soup = BeautifulSoup(response.text, "html.parser")

    for book in soup.select("article.product_pod"):
        title = book.h3.a["title"]

        price_text = book.select_one(".price_color").get_text(strip=True)
        price = float(price_text.replace("£", "").replace("Â", ""))

        rating_word = book.select_one(".star-rating")["class"][1]
        rating = rating_map[rating_word]

        availability = book.select_one(".availability").get_text(strip=True)

        relative_url = book.h3.a["href"]
        product_url = urljoin(url, relative_url)

        books.append({
            "Title": title,
            "Price": price,
            "Rating": rating,
            "Availability": availability,
            "URL": product_url
        })

    print(f"Page {page_number} scraped successfully.")

df = pd.DataFrame(books)

print("Total books scraped:", len(df))
display(df.head())
print(df.info())
```

```python
# Convert prices to pence to avoid decimal rounding issues
items = []

for _, book in df.iterrows():
    price_pence = round(book["Price"] * 100)

    if price_pence <= 10000:
        items.append({
            "Title": book["Title"],
            "Price": book["Price"],
            "Rating": int(book["Rating"]),
            "CostPence": price_pence
        })

# Dynamic programming: budget is 10,000 pence (£100)
# Each budget stores the best rating score and selected book indexes
dp = {0: (0, [])}

for i, book in enumerate(items):
    cost = book["CostPence"]
    rating = book["Rating"]

    new_dp = dp.copy()

    for spent, (score, selected) in dp.items():
        new_spent = spent + cost

        if new_spent <= 10000:
            new_score = score + rating

            if (
                new_spent not in new_dp
                or new_score > new_dp[new_spent][0]
            ):
                new_dp[new_spent] = (
                    new_score,
                    selected + [i]
                )

    dp = new_dp

# Choose the combination with the highest rating score.
# If scores tie, choose the lower-cost combination.
best_spent, (best_score, selected_indices) = max(
    dp.items(),
    key=lambda entry: (entry[1][0], -entry[0])
)

selected_books = pd.DataFrame(
    [items[i] for i in selected_indices]
)

print("CEO £100 Market-Research Challenge")
display(selected_books[["Title", "Price", "Rating"]])

print("Total cost: £", round(best_spent / 100, 2))
print("Budget remaining: £", round((10000 - best_spent) / 100, 2))
print("Total star-rating score:", best_score)
print("Number of books selected:", len(selected_books))
```
